```python
#1. Write a python function to Capitalize all words in a list of strings
string =["python","java","data analytics"]
def capitalize_all(string):
    return [a.title() for a in string]
capitalize_all(string)
```

```python
#3. Write a python function to Filter out strings that are palindromes
name="python"
def filter_palindrome(word):
    return[name==name [ : :-1] for i in word ]
filter_palindrome(name)
```

```python
#4. Write a python function to Concatenate all strings into one

list="python", "java", "data analytics"
def concatenate_str(word):
    return[list.join(list) for list in word]

concatenate_str(list)
```

```python
#5. Write a python function to Find the longest string in a list
new="the coders orbit"
def longest_str(lst):
    return[max(new) for new in lst]
longest_str(new)
```

```python
#10. Write a python function to Filter strings that start with a vowel

vowel = "python","apple","airline","indentation"
def filter_vowel(str):
    return[i.startswith("a,e,i,o,u") for i in str]
filter_vowel(vowel)
```
