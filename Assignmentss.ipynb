```python
#.You have a list of student scores: scores = [85, 92, 78, 65, 95, 70, -1, 88, 55, 100]. Iterate through the list:
#If a score is negative, consider it an invalid entry, print a warning message, and use continue to skip it.
#If a score is 100, print "Perfect Score!" and use break to stop further processing (assuming you only need to acknowledge the first perfect score).
#Otherwise, print the score and indicate if it's a passing score (70 or above) or a failing score


scores=[85,92,78,65,95,70,-1,88,55,100]
for score in scores :
    if score < 0:
        print("warning message: invalid entry")
        continue
        if score == 100:
            print("perfect score")
            break
            if score > 70:
                print("passing")
            else:
                print("failing")
print(scores)
```

```python
#write a python program with that defines user defined function to calculate the volume of cube and a sphere 

def volume_of_cube_sphere(side,radius):
    cube=  side **3
    sphere = (4/3)*3.141592*radius**3
    return  cube,sphere
print(volume_of_cube_sphere(4,2))
```

```python
#create a python program with user defined function to calculate the volume and surface area of a cylinder
def volume_surface_cylinder(radius,height):
    volume=3.141592 * radius**2 * height
    surface_area=2 * 3.141592 *radius * height + 2 *3.141592 * radius **2
    return volume,surface_area
print(volume_surface_cylinder(24,15))    
```

```python
#Write a Python program that defines a user-defined function calculate_hypotenuse(side1, side2) which takes the lengths of the two shorter sides (legs) of a right-angled triangle as input and returns the length of the hypotenuse.
def calculate_hypotenuse(side1,side2):
    length_of_hypotenuse=side1**2 + side2**2
    return length_of_hypotenuse
print(calculate_hypotenuse(5,6))
```
