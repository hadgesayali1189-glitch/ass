```python
#create a python program with user defined function to calculate the volume and surface area of a cylinder

def volume_cylinder(radius, height):
    volume= 3.141592 * (radius ** 2) * height
    return volume

def surface_cylinder(radius, height):
    surface=2 * 3.141592 * radius * (height + radius)
    return surface

print("volume of cylinder:", volume_cylinder(4,8))
print("surface area of cylinder:", surface_cylinder(12,31))
```
