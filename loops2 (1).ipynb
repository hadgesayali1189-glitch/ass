```python
# print cube of each number freom 1 to 5 using while loop

n = 1
while n <= 5:
    print(n ** 3)
    n += 1
```

```python
# for loop

n = 1
for n in range(1, 6):
    print(n ** 3)
```

```python
# print countdown frpm 20 to 10 using while loop

n = 20
while n >= 10:
    if n % 10 != 0:
      n -= 1
      print(n)
```

```python
# print all numbers between 1 to 50

n = 1
while n <= 50:
    n += 1
    print(n)
```

```python
# for loop


n = 1
for n in range(1, 51):
    if n % 10 != 0:
      print(n)
```

```python
#print sum of even numbers from 1 to 20 using while loop
even = 0
n = 1
while n <= 20:
    if n % 2 == 0:
       even += 1
       print(n)
```

```python
# for loop

even = 0
n = 1
for n in range(1, 21):
    if n % 2 == 0:
        even += 1
        n += 1
        print(n)
```
