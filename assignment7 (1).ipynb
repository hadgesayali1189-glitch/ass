```python
# write a function sales report...

def sales_report(region, *sales, **bonuses):
    print("region name", region)
    print("list of daily sales ammount", sales)
    for i in sales:
        print("sales are", i)
        print("keywords bonuses by day", bonuses)
        for k,v in bonuses.items():
            print("bonuses", k,v)
sales_report("East", 100, 250, 300, Day1 = 100, Day3 = 50) 
    
```

```python
#write a function account summary..

def account_summary(account_type, *transactions, **adjustments):
    print("account_type", account_type)
    print("transactions", transactions)
    for i in transactions:
        print("transactions are", i)
        print("adjustments", adjustments)
        for k,v in adjustments.items():
            print("adjustments", k,v)
account_summary("checking", 200, 600, -100, interest = 50, fees = 10)
```
