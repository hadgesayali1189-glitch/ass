```python
# write a function and returns both the count and sum of all factors of number

def factors_info(num):
    total = 0
    count = 0
    for i in range(1, num+1):
        if num % i == 0:
            count += 1
            total += i
    return count, total
print("factors for 12:", factors_info(12))
```
